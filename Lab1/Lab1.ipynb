Основні типи осередків (маркований список):
* **Code** — використовується для написання та виконання коду Python;
* **Markdown** — використовується для форматування тексту, формул та висновків.

### Пояснення роботи kernel
Після виконання **Restart Kernel** процес інтерпретатора Python перезапускається з нуля, тому вся оперативна пам'ять ядра очищується. Усі змінні, створені до перезапуску (зокрема змінна `x`), видаляються з пам'яті, і при спробі звернутися до `x` без повторного виконання клітинки її оголошення виникає помилка `NameError`.
